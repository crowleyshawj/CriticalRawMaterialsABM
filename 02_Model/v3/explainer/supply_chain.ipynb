# Mines

Firms will produce if their marginal cost to produce is equal to or less than the price forecasted. 

Mines will respond to two features:

    1. What is their forecast of the price for future periods?
    2. How much of their product was demanded in the previous period?

They will produce if their marginal cost is less than the forecast price, at the volume they expect to be traded. We take them to be somewhat simplistic initially, whereby they will adjust their production to the level that was traded in the previous level, if below full capacity.

# Markets

The price of the market is set as the marginal cost of the last unit traded. All firms will recieve that same price. 

Version 0 will set an artificially high price to start this process.
